```python
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

# Step 2: Assign weights and bias
w1 = 1
w2 = 1
b = -1.5

# Step 3: Step Activation Function
def step_function(z):
    if z >= 0:
        return 1
    else:
        return 0

# Step 4: Calculate output
print("AND Gate using Perceptron")
print("--------------------------")
print("x1  x2  Weighted Sum  Output")

for x1, x2 in inputs:

    # Calculate weighted sum
    z = (w1 * x1) + (w2 * x2) + b

    # Apply activation function
    output = step_function(z)

    # Display result
    print(x1, " ", x2, "     ", z, "       ", output)
```
