```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df
```

```python
print(df.dtypes)
```

```python
plt.subplot(2,2,1)
sns.histplot(df['total_bill'],bins = 20, kde = True)
plt.subplot(2,2,2)
sns.histplot(df['tip'],kde = True)
plt.subplot(2,2,3)
sns.histplot(df['size'],kde = True)

```
