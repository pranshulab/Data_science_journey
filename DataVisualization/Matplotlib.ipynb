```python
import numpy as np
import pandas as  pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.xlabel('x_axis')
plt.ylabel('y_axis')
plt.title('Demo')
```

Subplots in Matplpotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

**Object oriented way**

Matplotlib allow the aspect ratio, DPI and figure sixe to be specified when the figure object is created. you can use the figsixe and dpi keyword arguments

figsize is a tupple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch)

```python
fig = plt.figure(figsize = (8,8), dpi = 100)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.5,0.7,0.3,0.3])
axis2.plot(y,x)
```
