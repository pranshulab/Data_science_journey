```python
import numpy as np
import pandas as  pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.xlabel('x_axis')
plt.ylabel('y_axis')
plt.title('Demo')
```

Subplots in Matplpotlib

```python
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.subplot(2,2,2)
```
