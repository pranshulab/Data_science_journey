# Categotical Data Plots
now let's discus using seaborn to plot categorical data! There are few main plot types for this


*   boxplot

*   violinplot

*   stripplot

*   swarmplot

*   barplot

*   countplot







```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.countplot(x='sex',data=df, hue = 'smoker')
```

```python
sns.barplot(x='sex',y='total_bill',data=df)
```

You can change the estimator object to your own function, that converts a vector to a scalar

Mean (default): np.mean

Median: np.median

Sum: np.sum

Count: len (to count the number of observations)

 Standard Deviation: np.std

 Minimum: np.min


 Maximum: np.max

```python
import numpy as np
sns.barplot(x='sex',y='total_bill',data=df, estimator=np.sum)
```

**Boxplot**

```python
sns.boxplot(x='tip',y='day',data=df, palette='rainbow')
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = df)
```

```python
sns.stripplot(x='day',y='total_bill',data=df)
```

```python
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
sns.violinplot(x = 'tip', y = 'day' , data = df)
sns.swarmplot(x = 'tip', y = 'day' , data = df)
```
