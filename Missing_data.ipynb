```python
import numpy as np
import pandas as pd
```

**Finding missing Data**

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

Removing Missing data

```python
df
```

```python
df.dropna()
```

```python
df.dropna(thresh = 3)
```

Filling the missing data

```python
# fill all the missing value with zero
df.fillna(
    0
)
```

```python
# fill missing vslues with different values for different columns
values = {
      'A':0,
      'B':100,
      'C':300,
      'D':400
  }
df.fillna(value = values)
```

```python
df
```

```python
 # fill the vlaues with mean of all data
df.fillna(df.mean())
```
