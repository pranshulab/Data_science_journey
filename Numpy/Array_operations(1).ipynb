Deep and shallow copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a
```

```python
newa = a[4:6] * 10
newa
```

```python
b = a.copy()
```

```python
b[0] = 99
b
```
