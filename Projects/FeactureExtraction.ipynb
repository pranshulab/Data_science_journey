```python
def extract_episodes(txt):
  data = ""
  check = False
  for i in txt:
    if i == '(':
      check = True
      continue
    if i == ')':
        break
    if check:
      data += i
  return data
```

```python
df["epsodes"] = df["Title"].apply(extract_episodes)
```

```python
df["epsodes"] = df["epsodes"].str.replace(" eps", "")
```

```python
def extraction_time(txt):
  check = False
  data = ""
  for i in range(len(txt)):
    if txt[i] == ')':
      for j in range(i+1,i+20):
        data += txt[j]
      return data
```

```python
df['total time'] = df['Title'].apply(extraction_time)
```

```python
df.head()
```

```python
def total_months(x):
    start, end = x.split(" - ")

    start = pd.to_datetime(start, format="%b %Y")
    end = pd.to_datetime(end, format="%b %Y")

    return (end.year - start.year) * 12 + (end.month - start.month)

df["Months"] = df["total time"].apply(total_months)
```
