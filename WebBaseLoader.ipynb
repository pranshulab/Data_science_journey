```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline
```

```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
# downloading opensource model
llm = HuggingFacePipeline.from_model_id(
    model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    pipeline_kwargs= dict(
        temperature = 0.5,
        max_new_tokens = 100,
        do_sample = False,
        return_full_text = False
    )
)
```

```python
# from langchain_community.document_loaders import WebBaseLoader
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline


from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate



model = ChatHuggingFace(llm = llm)


prompt = PromptTemplate(
    template = """Answer the question using only the context below.
Keep your answer short and clear.
If the answer is not available, say so.

Question: {question}

Context:
{text}

Answer:""",
    input_variables = ['question', 'text']
)


parser = StrOutputParser()



# result = model.invoke("what is the capital of india")
# print(result.content)



url = 'https://www.flipkart.com/search?q=macbook+air+m5&as=on&as-show=on&otracker=AS_Query_OrganicAutoSuggest_3_6_na_na_na&otracker1=AS_Query_OrganicAutoSuggest_3_6_na_na_na&as-pos=3&as-type=RECENT&suggestionId=macbook+air+m5&requestId=4559e967-64d5-43c7-99c5-b2772e8f2446&as-searchtext=macbook%20air%20m5'

loader = WebBaseLoader(url)

docs = loader.load()


chain = prompt | model | parser

print(chain.invoke({
    'question': 'What is the prodcut that we are talking about?',
    'text':docs[0].page_content
}))
```
